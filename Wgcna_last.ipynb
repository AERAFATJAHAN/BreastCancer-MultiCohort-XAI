```python
import pandas as pd
import numpy as np

# 1. Load Datasets - Using the file that contains Gene_Symbol
df_names = pd.read_csv('/content/preprocessed_dataset_with_names.csv')
wgcna = pd.read_csv('/content/wgcna_gene_modules.csv')
deg_results = pd.read_csv('/content/deg_results_1685.csv')

# 2. Clean WGCNA Gene Names (Extract Symbol)
wgcna['Gene_Symbol'] = wgcna['Gene'].str.split('_').str[0]

# 3. Merge Expression Data with Modules
merged_full = pd.merge(df_names, wgcna[['Gene_Symbol', 'Module']], on='Gene_Symbol', how='inner')

# 4. Filter and Merge specifically for the 1685 DEGs
deg_final = pd.merge(deg_results, wgcna[['Gene_Symbol', 'Module']], on='Gene_Symbol', how='inner')

# 5. Analysis & Summary
print(f"Total Matched DEGs: {len(deg_final)}")
print("\nDEGs per Module:")
print(deg_final['Module'].value_counts())

# 6. Save Final Results
deg_final.to_csv('/content/final_deg_wgcna_analysis.csv', index=False)
print("\nFinal analysis saved to 'final_deg_wgcna_analysis.csv'")

display(deg_final.head())
```

### Module-Sample Relationship Heatmap
এই ধাপে আমরা মডিউল ইগেনজিন (Module Eigengenes) ব্যবহার করে স্যাম্পলগুলোর সাথে তাদের সম্পর্কের হিটম্যাপ তৈরি করব।

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Prepare expression data (Transpose so genes are columns)
# We'll use the merged_full which contains genes that were assigned to modules
expression_data = merged_full.drop(columns=['Entrez_Gene_Id', 'Module']).set_index('Gene_Symbol').T

# 2. Calculate Module Eigengenes (Simplified version: Mean expression per module)
# In real WGCNA this is the 1st Principal Component, but mean is a good approximation for visualization
modules = merged_full[['Gene_Symbol', 'Module']]
eigengenes = {}

for module in merged_full['Module'].unique():
    module_genes = modules[modules['Module'] == module]['Gene_Symbol']
    eigengenes[module] = expression_data[module_genes].mean(axis=1)

ME_df = pd.DataFrame(eigengenes)

# 3. Plot the Heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(ME_df.T, cmap='RdBu_r', center=0, cbar_kws={'label': 'Eigengene Expression'})
plt.title('WGCNA Module-Sample Relationship Heatmap')
plt.xlabel('Samples')
plt.ylabel('Modules')
plt.show()
```
